# ingesting data with autoloader

```sql
%python
dbutils.widgets.text("src","")
src = dbutils.widgets.get("src")
```

```sql
%python
df = spark.readStream.format("cloudFiles").option("cloudFiles.format", "csv")\
.option("cloudFiles.schemaLocation", f"/Volumes/airindia_flight_catalog/bronze/bronze_volume/{src}/checkpoint").option("cloudFiles.schemaEvolutionMode","rescue")\
.load(f"/Volumes/airindia_flight_catalog/raw/raw_volume/{src}/")
```

```sql
%python
df.writeStream\
  .format('delta')\
  .outputMode("append")\
  .trigger(once=True)\
  .option("checkpointLocation",f"/Volumes/airindia_flight_catalog/bronze/bronze_volume/{src}/checkpoint")\
  .option("path",f"/Volumes/airindia_flight_catalog/bronze/bronze_volume/{src}/data")\
  .start()
```
